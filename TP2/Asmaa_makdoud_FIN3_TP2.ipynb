```python
import pandas as pd

xls = pd.ExcelFile(chemin)
print('Feuilles :', xls.sheet_names)

pd.set_option('display.max_colwidth', None)
display(pd.read_excel(xls, sheet_name='Summary', header=None))
```

```python
df = pd.read_excel(xls, sheet_name='Data', header=0)

display(df.head(10))
print('Nombre de lignes :', df.shape[0])
print('Nombre de colonnes :', df.shape[1])
print(df.columns.tolist())
```

```python
import re

# 1) Renommer les deux colonnes aux noms très longs (par position)
cols = df.columns.tolist()
cols[0] = 'pays'
cols[2] = 'code_fmi'
df.columns = cols

# 2) Noms de colonnes en minuscules, sans espaces ni caractères spéciaux
def nettoyer(nom):
    nom = re.sub(r'[^0-9a-zA-Z_]+', '_', str(nom).strip().lower())
    return nom.strip('_')

df.columns = [nettoyer(c) for c in df.columns]

# 3) Convertir automatiquement en numérique les colonnes qui sont des nombres
for c in df.columns:
    if c in ('pays', 'iso3'):
        continue
    converti = pd.to_numeric(df[c], errors='coerce')
    n_orig = df[c].notna().sum()
    if n_orig > 0 and converti.notna().sum() >= 0.9 * n_orig:
        df[c] = converti

# 4) Supprimer les lignes sans pays ou sans année
df = df.dropna(subset=['pays', 'year'])
df['year'] = df['year'].astype(int)

# 5) Vérifier que les colonnes utilisées dans les requêtes existent
attendues = ['pays', 'year', 'iso3', 'gdp_lcu', 'totrev', 'taxrev', 'taxinc',
             'incomegroup', 'region', 'landlocked', 'fuelexporting',
             'resourcerich', 'smallisland']
absentes = [c for c in attendues if c not in df.columns]
print('Colonnes manquantes :', absentes if absentes else 'aucune')

df.info()
```

```python
print('Pays :', df['pays'].nunique(), '| Années :', df['year'].min(), '→', df['year'].max())
print('Doublons :', df.duplicated().sum())

manquants = df.isna().mean().mul(100).round(1).sort_values(ascending=False)
display(manquants.head(15))
```

```python
import sqlite3

conn = sqlite3.connect('tp_sql.db')
df.to_sql('donnees', conn, if_exists='replace', index=False)

def executer(titre, requete):
    """Affiche le titre puis le résultat de la requête SQL."""
    print('\n' + '=' * 70)
    print(titre)
    print('=' * 70)
    resultat = pd.read_sql_query(requete, conn)
    display(resultat)
    return resultat

executer('Vérification : 10 premières lignes', 'SELECT * FROM donnees LIMIT 10;')
executer('Nombre total de lignes importées', 'SELECT COUNT(*) AS total FROM donnees;')
```

```python
executer('1. Aperçu des colonnes principales', '''
SELECT pays, year, gdp_lcu, totrev, taxrev, taxinc
FROM donnees LIMIT 10;''')

executer('2. Nombre de pays distincts', '''
SELECT COUNT(DISTINCT pays) AS nb_pays FROM donnees;''')

executer('3. Plage des années', '''
SELECT MIN(year) AS annee_min, MAX(year) AS annee_max FROM donnees;''')

executer('4. Régions disponibles', '''
SELECT DISTINCT region FROM donnees ORDER BY region;''')

executer('5. Données du Maroc', '''
SELECT pays, year, gdp_lcu, totrev, taxrev
FROM donnees WHERE iso3 = 'MAR' ORDER BY year;''')
```

```python
executer('6. Les 10 plus fortes recettes fiscales', '''
SELECT pays, year, taxrev FROM donnees
WHERE taxrev IS NOT NULL
ORDER BY taxrev DESC LIMIT 10;''')

executer('7. Nombre d\'observations par région', '''
SELECT region, COUNT(*) AS observations
FROM donnees GROUP BY region ORDER BY observations DESC;''')

executer('8. Recettes fiscales moyennes par région', '''
SELECT region, ROUND(AVG(taxrev), 2) AS taxrev_moyen
FROM donnees WHERE taxrev IS NOT NULL
GROUP BY region ORDER BY taxrev_moyen DESC;''')

executer('9. Groupes de revenu avec plus de 500 observations de TaxRev', '''
SELECT incomegroup, COUNT(taxrev) AS nb_obs
FROM donnees
GROUP BY incomegroup HAVING COUNT(taxrev) > 500;''')

executer('10. Valeurs manquantes de TaxRev par région', '''
SELECT region,
       SUM(CASE WHEN taxrev IS NULL THEN 1 ELSE 0 END) AS taxrev_manquants,
       COUNT(*) AS total
FROM donnees GROUP BY region;''')
```

```python
executer('11. Pays dont le nom contient « Mor »', '''
SELECT DISTINCT pays FROM donnees WHERE pays LIKE '%Mor%';''')

executer('12. Pays enclavés entre 2015 et 2022', '''
SELECT pays, year, taxrev FROM donnees
WHERE year BETWEEN 2015 AND 2022
  AND landlocked = 'Landlocked'
  AND taxrev IS NOT NULL
ORDER BY pays, year LIMIT 15;''')

executer('13. Part fiscale dans les recettes totales (CASE WHEN)', '''
SELECT pays, year,
       ROUND(taxrev * 1.0 / totrev, 3) AS part_fiscale,
       CASE WHEN taxrev * 1.0 / totrev >= 0.8 THEN 'très fiscalisé'
            WHEN taxrev * 1.0 / totrev >= 0.5 THEN 'moyen'
            ELSE 'peu fiscalisé' END AS profil
FROM donnees
WHERE taxrev IS NOT NULL AND totrev IS NOT NULL AND totrev > 0
LIMIT 15;''')

executer('14. Sous-requête : TaxRev au-dessus de la moyenne', '''
SELECT pays, year, taxrev FROM donnees
WHERE taxrev > (SELECT AVG(taxrev) FROM donnees)
ORDER BY taxrev DESC LIMIT 10;''')

executer('15. Classement par région en 2020 (fonction fenêtre)', '''
SELECT region, pays, taxrev,
       RANK() OVER (PARTITION BY region ORDER BY taxrev DESC) AS rang
FROM donnees
WHERE year = 2020 AND taxrev IS NOT NULL
ORDER BY region, rang LIMIT 30;''')
```

```python
import matplotlib.pyplot as plt

# Équivalent Pandas de la requête 8
moy_region = (df.dropna(subset=['taxrev'])
                .groupby('region')['taxrev'].mean()
                .round(2).sort_values(ascending=False))
display(moy_region)

moy_region.sort_values().plot(kind='barh', figsize=(9, 4),
                              title='Recettes fiscales moyennes par région')
plt.xlabel('TaxRev moyen')
plt.tight_layout()
plt.show()

# Équivalent Pandas de la requête 5 : évolution du Maroc
maroc = df[df['iso3'] == 'MAR'].set_index('year')['taxrev']
maroc.plot(figsize=(9, 4), marker='o', title='Maroc : TaxRev au fil des années')
plt.ylabel('TaxRev')
plt.show()

# Équivalent Pandas de la requête 9 : observations par groupe de revenu
display(df.groupby('incomegroup')['taxrev'].count().sort_values(ascending=False))
```
